**1. Architecture reminders and intuition**


**In one or two sentences, explain how attention in a Transformer for sequences differs from
attention in a graph model.**       

In a Transformer, self-attention treats a sequence as a fully connected graph where every token can attend to all other tokens regardless of distance. In a graph model (such as a Graph Attention Network), attention is typically restricted to local neighborhoods defined by explicit, pre-defined edges in the graph topology rather than all-to-all connectivity

**Why can the positional encodings used for sequences not be applied directly to arbitrary graphs?**        

Standard sequence positional encodings cannot be applied directly to arbitrary graphs because graphs have no natural linear order, varying node counts, and complex structural topologies. While text sequences follow a strict one-dimensional path from left to right, graph nodes are unordered sets connected by arbitrary edges.


**Give two potential advantages of attention over graph neighborhoods.**        

Two potential advantages of attention mechanisms (such as in Graph Attention Networks) over fixed or uniform graph neighborhood aggregation (like standard Graph Convolutional Networks) are:       

• Data-dependent, adaptive weighting: Traditional graph neighborhoods treat all connected neighbors equally or rely strictly on static topological features like node degree. Attention dynamically assigns learnable, content-based importance scores to each neighbor depending on their actual feature values, allowing the model to focus on the most relevant connections.     

• Multi-head representation capacity: Attention mechanisms can utilize multiple "heads" to independently capture different types of relationships or semantic/structural aspects simultaneously (e.g., one head tracking role similarity and another tracking local cluster proximity), whereas standard neighborhood aggregation blends all neighbor information into a single homogeneous representation
